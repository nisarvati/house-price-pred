Data Preprocessing

```python
#separate features and target variable
X = df.drop(columns = [TARGET])
y = df[TARGET]
```

```python
X.head()
```

```python
y.head()
```

```python
#train test split
X_train, X_test, y_train, y_test = train_test_split(
  X, y, test_size = 0.2, random_state = RANDOM_STATE
)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
numerical_features = X_train.select_dtypes(include=[np.number]).columns.tolist()
categorical_features = X_train.select_dtypes(exclude=[np.number]).columns.tolist()
print("Numerical Features: ", numerical_features)
print("Categorical Features: ",categorical_features)
```

```python
numerical_features
```

```python
#numerical features preprocessing steps
numerical_transformer = Pipeline(steps=[
  ('imputer', SimpleImputer(strategy='median')),
  ('scaler', StandardScaler())
])

#preprocessing steps for categorical features
categorical_transformer = Pipeline(steps=[
  ('imputer', SimpleImputer(strategy='most_frequent')),
  ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

#combine preprocessing steps, automate the preprocessing
preprocess = ColumnTransformer(
  transformers=[
    ('num', numerical_transformer, numerical_features),
    ('cat', categorical_transformer, categorical_features)
  ]
)
```

Baseline Model (no CV, no Tuning)

```python
baseline_pipe = Pipeline(
  steps = [
    ("preprocess", preprocess),
    ("model", LinearRegression())
  ]
)
```

```python
baseline_pipe.fit(X_train, y_train)
```

```python
train_baseline_pred = baseline_pipe.predict(X_train)
test_baseline_pred = baseline_pipe.predict(X_test)
```

```python
#first 5 predictions on training set
train_baseline_pred[:5]
```

```python
y_train[:5]
```

```python
#errors on training set
train_baseline_rmse = root_mean_squared_error(y_train, train_baseline_pred)
train_baseline_mae = mean_absolute_error(y_train, train_baseline_pred)
train_baseline_r2 = r2_score(y_train, train_baseline_pred)

print("Training Set Errors:")
print(f"RMSE: {train_baseline_rmse:.2f}")
print(f"MAE: {train_baseline_mae:.2f}")
print(f"R2 Score: {train_baseline_r2:.4f}")
```

```python
#errors on test set
test_baseline_rmse = root_mean_squared_error(y_test, test_baseline_pred)
test_baseline_mae = mean_absolute_error(y_test, test_baseline_pred)
test_baseline_r2 = r2_score(y_test, test_baseline_pred)

print("Training Set Errors:")
print(f"RMSE: {test_baseline_rmse:.2f}")
print(f"MAE: {test_baseline_mae:.2f}")
print(f"R2 Score: {test_baseline_r2:.4f}")

#rmse depends on the scale of target variable, so it is better to compare mae values
```

Model Selection and Optimization

```python
#selecting different models to compare with baseline model

models = {
  "LinearRegression": LinearRegression(),
  "Ridge": Ridge(random_state=RANDOM_STATE),
  "Lasso": Lasso(random_state=RANDOM_STATE, max_iter=10000),
  "RandomForest": RandomForestRegressor(),
  "HistGB" : HistGradientBoostingRegressor()
}
```

```python
k = 5
cv = KFold(n_splits=k, shuffle=True, random_state=RANDOM_STATE)
```

```python
scoring = {
  "rmse" : "neg_root_mean_squared_error",
  "mae" : "neg_mean_absolute_error",
  "r2" : "r2"
}
```

```python
rows = []

for name, model in models.items():
  pipe = Pipeline(
    steps = [
      ("preprocess", preprocess),
      ("model", model)
    ]
  )

  scores = cross_validate(pipe, X_train, y_train, cv = cv, scoring = scoring, n_jobs = 1)
  rows.append(
    {
      "model": name,
      "cv_rmse": -scores["test_rmse"].mean(),
      "cv_mae": -scores["test_mae"].mean(),
      "cv_r2": scores["test_r2"].mean() 
    }
  )

cv_results = pd.DataFrame(rows).sort_values("cv_rmse")
cv_results
```

```python
scores
```

```python
rows
```

```python
best_row = cv_results.sort_values("cv_rmse").iloc[0]

best_model_name = best_row["model"]
best_rmse = best_row["cv_rmse"]

print(f"Best Model: {best_model_name} with CV RMSE: {best_rmse:.2f}")
```

Best Model: HistGradientBoostingRegressor

Hyper Parameter Tuning

```python
#hyperparameters are epochs and learning rate
hgb_pipe = Pipeline(
  steps = [
    ("preprocess", preprocess),
    ("model", HistGradientBoostingRegressor(random_state=RANDOM_STATE))
  ]
)
```

```python
#hyperparameters combination
param_grid = {
  #model is added because we are tuning the model step in the pipeline, not the preprocess step
  "model__learning_rate": [0.03,0.05, 0.1],
  "model__max_depth": [None, 3,6],
  "model__max_leaf_nodes": [15,31,63],
  "model__min_samples_leaf": [20,50,100],
  "model__l2_regularization": [0, 0.1, 1]
}
```

```python
grid = GridSearchCV(
  estimator=hgb_pipe,
  param_grid=param_grid,
  cv=cv,
  scoring="neg_root_mean_squared_error",
  n_jobs=-1,
  verbose=1
  #retraining can also be done by adding refit=True, and then call grid.best_estimator_
)
```

```python
#perform grid search 
grid.fit(X_train,y_train)
```

```python
print("Tuned HistGB: (CV)")
print("Best CV RMSE: ", -grid.best_score_)
print("Best params: ", grid.best_params_)
```

Retraining with best params

```python
grid.best_estimator_
```

```python
#can also be done by refit=True in gridsearch
hgb_best = Pipeline(
  steps = [
    ("preprocess", preprocess),
    ("model", HistGradientBoostingRegressor(
        l2_regularization=0.1,
        learning_rate=0.1,
        max_depth=None,
        max_leaf_nodes=63,
        min_samples_leaf=20
    ))
  ]
)
```

```python
hgb_best.fit(X_train,y_train)
```

Final Evaluation

```python
train_final_pred = hgb_best.predict(X_train)

train_final_rmse = root_mean_squared_error(y_train, train_final_pred)
train_final_mae = mean_absolute_error(y_train, train_final_pred)
train_final_r2 = r2_score(y_train, train_final_pred)

print("Final Training Set Errors:")
print(f"RMSE: {train_final_rmse:.3f}")
print(f"MAE: {train_final_mae:.3f}")
print(f"R2 Score: {train_final_r2:.3f}")
```

```python
test_final_pred = hgb_best.predict(X_test)

test_final_rmse = root_mean_squared_error(y_test, test_final_pred)
test_final_mae = mean_absolute_error(y_test, test_final_pred)
test_final_r2 = r2_score(y_test, test_final_pred)

print("Final Test Set Errors:")
print(f"RMSE: {test_final_rmse:.3f}")
print(f"MAE: {test_final_mae:.3f}")
print(f"R2 Score: {test_final_r2:.3f}")
```

Build a predictive System

```python
def predict_house_price(
    model,
    longitude: float,
    latitude: float,
    housing_median_age: float,
    total_rooms: float,
    total_bedrooms: float,
    population: float,
    households: float,
    median_income: float,
    ocean_proximity: str
) -> float:
    """
    Predict median_house_value for one new house.
    total_bedrooms can be np.nan (pipeline will impute).
    """
    new_row = pd.DataFrame([{
        "longitude": longitude,
        "latitude": latitude,
        "housing_median_age": housing_median_age,
        "total_rooms": total_rooms,
        "total_bedrooms": total_bedrooms,
        "population": population,
        "households": households,
        "median_income": median_income,
        "ocean_proximity": ocean_proximity
    }])

    return float(model.predict(new_row)[0])
```

```python
example_pred = predict_house_price(
    model=hgb_best,
    longitude=-122.230,
    latitude=37.880,
    housing_median_age=41,
    total_rooms=880,
    total_bedrooms=129,
    population=322,
    households=126,
    median_income=8.3252,
    ocean_proximity="NEAR BAY"
 )

print("\nExample prediction:", round(example_pred, 2))
```
